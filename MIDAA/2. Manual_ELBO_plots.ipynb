The number of archetypes dictates the number of vertices of the resulting polytope. The ideal model balances performance on ELBO score and a low number of vertices to maintain biolgical interpretability. This can be identified using the elbow of final ELBO plots generated in pipeline 1. This happens automatically on the MIDAA_pipeline notebook, but also put here in case its helpful to track progress before all the range of archetypes have been built.

## 1. Load data and models
Loading the adata object and pretrained models 

```python
import torch
import numpy as np
import scanpy as sc
import matplotlib.pyplot as plt
from scipy.sparse import issparse
import os
import pickle
```

```python
# Load AnnData object
adata_path = r"C../Data" #update to be your anndata file
adata = sc.read_h5ad(adata_path)
print(f'Loaded AnnData with shape: {adata.shape}') #QC
```

```python
#loading pretrained models --> update with yours (path numbered correspond to number of archetypes) (comment out the ones not loaded yet)
path2 = r"XX"
path3 = r"XX"
path4 = r"XX"
path5 = r"XX"
path6 = r"XX"
path7 = r"XX"
path8 = r"XX"
path9 = r"XX"
path10 = r"XX"
path11 = r"XX"


# Dict mapping number of archetypes to its model path (comment out/remove the ones not loaded yet)
model_paths = {
    2: path2, 3: path3, 4: path4,
    5: path5, 6: path6,
    7: path7, 8: path8, 9: path9, 10: path10, 11: path11,
}

# For each model path, load the final ELBO value, keyed by its actual archetype number

ELBO_by_narch = {}

for narchetypes, path in model_paths.items():
    with open(path, "rb") as f: #read binary, giving opened file temporary name "f"
        result = pickle.load(f)
        ELBO_by_narch[narchetypes] = result["ELBO"]
```

Plot final elbo values on a graph to identify the best model performance

```python
import matplotlib.pyplot as plt

archetype_nums = sorted(ELBO_by_narch.keys()) #uses the actual archetype numbers, so this stays correct even if some paths are skipped
final_elbos = [ELBO_by_narch[n][-1] if isinstance(ELBO_by_narch[n], (list, np.ndarray)) else ELBO_by_narch[n] for n in archetype_nums] #accessing the last value of each elbo list

plt.figure(figsize=(8, 5))
plt.plot(archetype_nums, final_elbos, marker='o')
plt.xlabel('Number of archetypes')
plt.ylabel('Final ELBO value')
plt.title('Final ELBO value vs Number of Archetypes')
plt.show()

```
